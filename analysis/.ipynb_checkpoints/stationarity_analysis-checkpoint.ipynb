# Stationarity test results

Exploratory analysis of the CSVs produced by `experiments/stationarity_tests.py`.

**Files expected:**
- `../experiments/stationarity_results/summary.csv`
- `../experiments/stationarity_results/per_node.csv`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

RESULTS_DIR = "../experiments/stationarity_results"
ALPHA_ADF   = 0.05   # p < α → ADF calls stationary
ALPHA_KPSS  = 0.05   # p < α → KPSS calls non-stationary

summary  = pd.read_csv(f"{RESULTS_DIR}/summary.csv")
per_node = pd.read_csv(f"{RESULTS_DIR}/per_node.csv")

print(f"Datasets : {len(summary)}")
print(f"Node rows: {len(per_node)}")
```

## Summary table

```python
display_cols = [
    "dataset_label", "generator", "scenario", "T", "N",
    "adf_overall_call", "kpss_overall_call",
    "delta_mean_mean", "delta_var_pct_mean", "delta_corr",
]
summary[display_cols].style \
    .set_caption("One row per dataset — overall stationarity verdicts and structural-change metrics") \
    .format({
        "delta_mean_mean":  "{:.3f}",
        "delta_var_pct_mean": "{:+.1f}%",
        "delta_corr":       "{:.3f}",
    }) \
    .background_gradient(subset=["delta_corr"], cmap="YlOrRd")
```

## Per-node ADF p-values  
Colour: green = stationary (p < 0.05), red = non-stationary.  
Each row is one dataset, each column one node.

```python
def pvalue_heatmap(per_node, col_pval, col_threshold, call_direction,
                   title, alpha=0.05, figsize=(10, 0.6)):
    """
    col_pval       : column name for p-value
    call_direction : 'low_is_detected' (ADF: p<α → stationary, detected)
                     'low_is_nonstat'  (KPSS: p<α → non-stationary)
    """
    pivot = per_node.pivot_table(
        index="dataset_label", columns="node", values=col_pval, aggfunc="first"
    )
    datasets = pivot.index.tolist()
    n_nodes  = pivot.shape[1]

    fig, ax = plt.subplots(figsize=(max(6, n_nodes * 1.1), max(2, len(datasets) * figsize[1])))

    # Colour: green where test "passes" (stationary / not-flagged), red where it doesn't
    if call_direction == "low_is_detected":
        # ADF: low p = stationary = green
        colour_arr = np.where(pivot.values < alpha, 0.9, 0.1)   # 1=green, 0=red in RdYlGn
        cmap = "RdYlGn"
    else:
        # KPSS: low p = NON-stationary = red
        colour_arr = np.where(pivot.values < alpha, 0.1, 0.9)
        cmap = "RdYlGn"

    im = ax.imshow(colour_arr, cmap=cmap, vmin=0, vmax=1, aspect="auto")

    # Annotate with p-values
    for i, ds in enumerate(datasets):
        for j, node in enumerate(pivot.columns):
            pv = pivot.loc[ds, node]
            if pd.notna(pv):
                ax.text(j, i, f"{pv:.3f}", ha="center", va="center", fontsize=8)

    ax.set_xticks(range(n_nodes))
    ax.set_xticklabels([f"node {c}" for c in pivot.columns], fontsize=8)
    ax.set_yticks(range(len(datasets)))
    ax.set_yticklabels(datasets, fontsize=8)
    ax.set_title(title, fontsize=10)

    # Significance threshold line legend
    from matplotlib.patches import Patch
    if call_direction == "low_is_detected":
        patches = [
            Patch(color=plt.cm.RdYlGn(0.9), label=f"p < {alpha}  → stationary"),
            Patch(color=plt.cm.RdYlGn(0.1), label=f"p ≥ {alpha}  → non-stationary"),
        ]
    else:
        patches = [
            Patch(color=plt.cm.RdYlGn(0.1), label=f"p < {alpha}  → non-stationary (flagged)"),
            Patch(color=plt.cm.RdYlGn(0.9), label=f"p ≥ {alpha}  → stationary (not flagged)"),
        ]
    ax.legend(handles=patches, loc="upper right", bbox_to_anchor=(1.35, 1.0), fontsize=8)
    plt.tight_layout()
    plt.show()

pvalue_heatmap(per_node, "adf_pvalue",  "adf_call",  "low_is_detected",
               f"ADF p-values per node  (green = stationary, α={ALPHA_ADF})")
pvalue_heatmap(per_node, "kpss_pvalue", "kpss_call", "low_is_nonstat",
               f"KPSS p-values per node  (red = non-stationary flagged, α={ALPHA_KPSS})")
```

## Test sensitivity per dataset
For each dataset: what fraction of nodes does each test flag as non-stationary?  
High bar = test detects something.  For `modular_to_hub` we expect all bars ≈ 0 except ΔCorr.

```python
# Fraction of nodes flagged as non-stationary by each test
def frac_flagged(group):
    n = len(group)
    return pd.Series({
        "ADF: frac non-stationary":  (group["adf_pvalue"]  >= ALPHA_ADF).sum()  / n,
        "KPSS: frac non-stationary": (group["kpss_pvalue"] < ALPHA_KPSS).sum() / n,
        "Δmean (mean)": group["delta_mean"].mean(),
        "|Δvar%| (mean)": group["delta_var_pct"].abs().mean(),
    })

node_flags = per_node.groupby("dataset_label").apply(frac_flagged)

# Add ΔCorr from summary (multivariate, one value per dataset)
node_flags = node_flags.join(
    summary.set_index("dataset_label")[["delta_corr"]]
)

# Normalise ΔCorr to [0,1] range for comparison (divide by max)
dc_max = node_flags["delta_corr"].max()
node_flags["ΔCorr (normalised)"] = node_flags["delta_corr"] / (dc_max + 1e-9)

plot_cols = [
    "ADF: frac non-stationary",
    "KPSS: frac non-stationary",
    "Δmean (mean)",
    "|Δvar%| (mean)",
    "ΔCorr (normalised)",
]

ax = node_flags[plot_cols].plot(
    kind="bar", figsize=(12, 4), width=0.75,
    color=["#e07b54", "#5c8ecc", "#6abf69", "#f2c94c", "#9b59b6"]
)
ax.set_title("Detection sensitivity per dataset and test", fontsize=11)
ax.set_xlabel("")
ax.set_ylabel("Fraction flagged / normalised metric", fontsize=9)
ax.set_xticklabels(ax.get_xticklabels(), rotation=25, ha="right", fontsize=8)
ax.axhline(0.5, color="black", lw=0.8, linestyle=":", label="50% threshold")
ax.legend(fontsize=8, loc="upper right")
plt.tight_layout()
plt.show()

node_flags[plot_cols + ["delta_corr"]].round(3)
```

## Δmean and Δvar distributions (per node, per dataset)

```python
datasets = per_node["dataset_label"].unique()
n_ds = len(datasets)

fig, axes = plt.subplots(1, 2, figsize=(12, max(3, n_ds * 0.5 + 1)))

for ax, col, xlabel in [
    (axes[0], "delta_mean",    "Standardised mean shift  |Δμ|/σ"),
    (axes[1], "delta_var_pct", "Relative variance change  Δσ (%)"),
]:
    for i, ds in enumerate(datasets):
        vals = per_node.loc[per_node["dataset_label"] == ds, col].dropna()
        ax.scatter(vals, [i] * len(vals), s=40, alpha=0.8, label=ds)
        ax.scatter(vals.mean(), i, marker="|", s=200, color="black", zorder=5)

    ax.set_yticks(range(n_ds))
    ax.set_yticklabels(datasets, fontsize=8)
    ax.set_xlabel(xlabel, fontsize=9)
    ax.axvline(0, color="grey", lw=0.8, linestyle="--")
    ax.grid(axis="x", lw=0.5, alpha=0.4)

fig.suptitle("Per-node structural-change metrics (dots = nodes, bar = mean)", fontsize=10)
plt.tight_layout()
plt.show()
```

## ΔCorr vs Δmean — the blind spot of marginal tests
Top-right: connectivity changed AND marginal mean shifted (easy case).  
Top-left: connectivity changed but marginals unchanged (hard case — `modular_to_hub`).  
Bottom-right: mean shifted but connectivity unchanged.

```python
fig, ax = plt.subplots(figsize=(7, 5))

for _, row in summary.iterrows():
    ax.scatter(row["delta_mean_mean"], row["delta_corr"],
               s=120, zorder=3)
    ax.annotate(row["dataset_label"],
                (row["delta_mean_mean"], row["delta_corr"]),
                textcoords="offset points", xytext=(6, 4), fontsize=7)

ax.set_xlabel("Δmean (mean across nodes)", fontsize=9)
ax.set_ylabel("ΔCorr (Frobenius, correlation matrices)", fontsize=9)
ax.set_title("Structural change: marginal (Δmean) vs network (ΔCorr)", fontsize=10)
ax.axhline(0, color="grey", lw=0.5)
ax.axvline(0, color="grey", lw=0.5)

# Quadrant labels
xlim = ax.get_xlim(); ylim = ax.get_ylim()
mx, my = (xlim[0] + xlim[1]) / 2, (ylim[0] + ylim[1]) / 2
for txt, x, y in [
    ("marginals+network\nchanged",   xlim[1]*0.6, ylim[1]*0.85),
    ("network changed\nmarginals ok",xlim[0]*0.05, ylim[1]*0.85),
    ("marginals changed\nnetwork ok",xlim[1]*0.6, ylim[0]*0.15 + ylim[0]),
]:
    ax.text(x, y, txt, fontsize=7, color="grey", ha="center")

plt.tight_layout()
plt.show()
```

## ADF p-value distribution per dataset
Box plots across nodes — shows whether unit-root tests are at floor (stationary) or elevated.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4), sharey=False)

for ax, col, title, alpha_line, low_bad in [
    (axes[0], "adf_pvalue",  "ADF p-values per dataset\n(high p = non-stationary)",  ALPHA_ADF,  False),
    (axes[1], "kpss_pvalue", "KPSS p-values per dataset\n(low p = non-stationary)",  ALPHA_KPSS, True),
]:
    groups  = [per_node.loc[per_node["dataset_label"] == ds, col].dropna().values
               for ds in datasets]
    bp = ax.boxplot(groups, vert=True, patch_artist=True,
                    medianprops=dict(color="black", lw=1.5))
    for patch in bp["boxes"]:
        patch.set_facecolor("#aec6e8")

    ax.axhline(alpha_line, color="crimson", lw=1.2, linestyle="--",
               label=f"α = {alpha_line}")
    ax.set_xticks(range(1, len(datasets) + 1))
    ax.set_xticklabels(datasets, rotation=30, ha="right", fontsize=7)
    ax.set_ylabel("p-value", fontsize=9)
    ax.set_title(title, fontsize=9)
    ax.legend(fontsize=8)
    ax.set_ylim(-0.02, 1.05)

plt.tight_layout()
plt.show()
```

## Raw per-node table  
Full per-node CSV for inspection and filtering.

```python
# Filter to a specific dataset if needed — set to None for all
FILTER_DATASET = None  # e.g. "var_modular_to_hub"

df = per_node if FILTER_DATASET is None else per_node[per_node["dataset_label"] == FILTER_DATASET]

show_cols = ["dataset_label", "node",
             "adf_pvalue", "adf_call",
             "kpss_pvalue", "kpss_call",
             "delta_mean", "delta_var_pct"]

def _colour_call(val):
    if val == "stationary":     return "background-color: #c8f7c5"
    if val == "non-stationary": return "background-color: #f7c5c5"
    return ""

df[show_cols].style \
    .applymap(_colour_call, subset=["adf_call", "kpss_call"]) \
    .format({
        "adf_pvalue":   "{:.4f}",
        "kpss_pvalue":  "{:.4f}",
        "delta_mean":   "{:.4f}",
        "delta_var_pct":"{:+.2f}%",
    })
```
